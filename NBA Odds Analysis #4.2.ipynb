```python
import pandas as pd
pd.set_option('display.max_columns', None)

with open('games_data.json', 'r') as file:
    data = json.load(file)

#print(data)
pretty_data = json.dumps(data, indent=4)
print(pretty_data)
```

```python
from datetime import datetime
import pytz

extracted_data = []
utc_zone = pytz.timezone('UTC')
eastern_zone = pytz.timezone('US/Eastern')
# Loop through each entry in the JSON
for entry in data:
    match_id = entry['id']
    home_team = entry['home_team']
    away_team = entry['away_team']
    startDateTime = entry['commence_time']
    utc_time = datetime.strptime(startDateTime, "%Y-%m-%dT%H:%M:%SZ")
    utc_time = utc_zone.localize(utc_time)
    eastern_time = utc_time.astimezone(eastern_zone)
    eastern_time.strftime('%Y-%m-%d %H:%M:%S %Z')

    for bookmaker in entry['bookmakers']:
        bookmaker_title = bookmaker['title']
        
        # Loop through markets and find where market key is 'h2h'
        for market in bookmaker['markets']:
            if market['key'] == 'h2h':
                for outcome in market['outcomes']:
                    extracted_data.append({
                        'ID': match_id,
                        'StartDate': eastern_time,
                        'HomeTeam': home_team,
                        'AwayTeam': away_team,
                        'Bookmaker': bookmaker_title,
                        'OutcomeTeam': outcome['name'],
                        'Price': outcome['price']
                    })

# Create DataFrame
df = pd.DataFrame(extracted_data)

# Display DataFrame
print(df)
```

```python
df['adjusted_price'] = ((df['Price'] * 1) - 1)
```

```python
def decimal_to_us_odds(decimal_odds):
    if decimal_odds == 1.00:
        return 0  # No payout, return 0 or another meaningful value
    elif decimal_odds >= 2:
        # Underdogs: (decimal_odds - 1) * 100
        return (decimal_odds - 1) * 100
    else:
        # Favorites: -100 / (decimal_odds - 1)
        return -100 / (decimal_odds - 1)
```

```python
def extract_date(datetime):
    # Parse the datetime string
    dt = datetime.date()
    time = datetime.time()
    # Extract the date part
    return dt

def extract_time(datetime):
    # Parse the datetime string
    time = datetime.time()
    # Extract the date part
    return time
```

```python
df_pivot = pd.pivot_table(maxvaluedf, 
                          index=['gameid', 'date'], 
                          columns=maxvaluedf.groupby('gameid').cumcount() + 1, 
                          values=['Team', 'Highest_US_Odds', 'decimalOdds','Sportsbook'], 
                          aggfunc='first')

# Flatten the columns to make them more readable
df_pivot.columns = [f'{col[0]}{col[1]}' for col in df_pivot.columns]

# Reset index for cleaner output
df_pivot = df_pivot.reset_index()

# Display the pivoted dataframe
df_pivot['stake'] = 50
df_pivot
```

```python
def implied_probability(decimal_odds):
    if decimal_odds <= 1:  # Handle non-positive odds case
        return None
    return 1 / decimal_odds

def calculate_arbitrage(row):
    prob1 = implied_probability(row['decimalOdds1'])
    prob2 = implied_probability(row['decimalOdds2'])
    
    # Skip if one of the probabilities is None
    if prob1 is None or prob2 is None:
        return pd.Series({
            "Arbitrage_Opportunity": False,
            "Arbitrage_Percentage": None,
            "prob1": None,
            "prob2": None,            
            "Stake_on_Team1": None,
            "Stake_on_Team2": None
        })

    total_prob = prob1 + prob2
    
    if total_prob < 1:
        arbitrage_percentage = 1 - total_prob
        stake1 = (row['stake'] * prob2) / total_prob
        stake2 = (row['stake'] * prob1) / total_prob
        return pd.Series({
            "Arbitrage_Opportunity": True,
            "Arbitrage_Percentage": arbitrage_percentage,
            "prob1": prob1,
            "prob2": prob2,
            "Stake_on_Team1": stake1,
            "Stake_on_Team2": stake2
        })
    else:
        return pd.Series({
            "Arbitrage_Opportunity": False,
            "Arbitrage_Percentage": None,
            "prob1": None,
            "prob2": None,            
            "Stake_on_Team1": None,
            "Stake_on_Team2": None
        })

# Apply the function to each row
arbitrage_results = df_pivot.apply(calculate_arbitrage, axis=1)

# Concatenate results with original DataFrame
result_df = pd.concat([df_pivot, arbitrage_results], axis=1)

# Display the result DataFrame
result_df
```
